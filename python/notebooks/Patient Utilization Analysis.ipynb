What types of encounters are driving high-utilizer patients?

```python
encounters.head()
```

```python
high_utilizer_ids = patient_utilization.loc[
    patient_utilization['high_utilizer'],
    'patient_id'
]

high_utilizer_ids.head()    #now have high_utilizer_ids, containing the patient IDs of your 1,199 high-utilizer patients.
```

```python
high_utilizer_encounters = encounters[
    encounters['PATIENT'].isin(high_utilizer_ids)
]

high_utilizer_encounters.shape
```

What types of encounters are most common among high-utilizer patients?

```python
high_utilizer_encounters['ENCOUNTERCLASS'].value_counts()
```

High utilization in this dataset is not necessarily driven by frequent hospitalization or emergency visits. A large portion comes from repeated routine/outpatient healthcare interactions.

Convert these counts into percentages

```python
high_utilizer_encounter_pct = (
    high_utilizer_encounters['ENCOUNTERCLASS']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)                                     #.value_counts(normalize=True)  #would give:

#wellness       0.50
#outpatient     0.30
#emergency      0.20       #.mul(100) means:

                           #multiply every value by 100
                                    

high_utilizer_encounter_pct
```

```python
high_utilizer_encounter_pct.sort_values(ascending=False)
```

Which encounter types are disproportionately associated with high-utilizer patients?

We want to compare high-utilizer encounter distribution against the overall encounter distribution you calculated earlier.

```python
overall_encounter_pct = (
    encounters['ENCOUNTERCLASS']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

encounter_comparison = pd.DataFrame({
    'overall_pct': overall_encounter_pct,
    'high_utilizer_pct': high_utilizer_encounter_pct
})

encounter_comparison['difference'] = (
    encounter_comparison['high_utilizer_pct']
    - encounter_comparison['overall_pct']
)

encounter_comparison.sort_values(                                 #This answers:

                                                                  #Which encounter types are more prominent among high-utilizer patients than in the overall population?
    'difference',
    ascending=False
)
```

Interpretation: High-utilizer patients have a substantially higher share of outpatient encounters than the overall population (31.43% vs 18.15%, +13.28 percentage points)
