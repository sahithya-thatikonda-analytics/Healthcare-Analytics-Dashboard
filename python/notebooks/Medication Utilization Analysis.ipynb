Next: calculate the IQR

The IQR (Interquartile Range) measures the spread of the middle 50%:

IQR=Q3−Q1

```python
Q1 = patient_medication_burden['total_medication_cost'].quantile(0.25)
Q3 = patient_medication_burden['total_medication_cost'].quantile(0.75)

IQR = Q3 - Q1

Q1, Q3, IQR
```

```python
upper_bound = Q3 + (1.5 * IQR)

upper_bound   #This means patients with medication costs above ~386,737 will be considered unusually high-cost patients using the standard IQR outlier rule.
```

Cost concentration / Pareto analysis

Business question:

Are medication costs concentrated among a small proportion of patients?

Step 1 — Sort patients by medication cost

```python
patient_medication_burden.head(2)
```

```python
patient_medication_burden.columns.tolist()
```

```python
cost_concentration = (
    patient_medication_burden
    .reset_index()
    [['PATIENT', 'total_medication_cost']]
    .sort_values('total_medication_cost', ascending=False)
    .reset_index(drop=True)
)

cost_concentration.head(10)
```

Step 2 — Calculate cumulative spending

```python
cost_concentration['cumulative_cost'] = (
    cost_concentration['total_medication_cost'].cumsum()
)

total_cost = cost_concentration['total_medication_cost'].sum()

cost_concentration['cumulative_cost_pct'] = (
    cost_concentration['cumulative_cost'] / total_cost * 100
).round(2)

cost_concentration.head(10)
```

Step 3 — Find how much of total spending comes from the top 10% of patients

```python
top_10_pct_count = int(len(cost_concentration) * 0.10)                                           #9,566 patients.

                                                                                             #So:9566×10%=956.6

top_10_cost_pct = (
    cost_concentration.iloc[:top_10_pct_count]['total_medication_cost'].sum()
    / total_cost * 100
)

round(top_10_cost_pct, 2)
```

Total patients: 9,566
Top 10% ≈ 956 patients
These 956 highest-cost patients account for 44.26% of total medication spending

So the key finding is:

The top 10% of patients by medication cost account for 44.26% of the total medication spending.

That's a strong indication of cost concentration: a relatively small group of patients contributes a disproportionately large share of medication expenditure.

---What percentage of spending is generated by the top 10% vs top 20% of patients?

```python
top_20_pct_count = int(len(cost_concentration) * 0.20)

top_20_cost_pct = (
    cost_concentration.iloc[:top_20_pct_count]['total_medication_cost'].sum()
    / total_cost * 100
)

round(top_20_cost_pct, 2)
```

This is a strong cost-concentration finding.

Just 10% of patients generate 44.26% of total medication spending.
Expanding to the top 20%, they account for 64.21% of spending.
Therefore, the remaining 80% of patients account for only 35.79% of spending.

Medication expenditure was highly concentrated among a relatively small proportion of patients. The top 10% of patients by medication cost accounted for 44.26% of total medication expenditure, while the top 20% accounted for 64.21%.

---How many patients in each medication-burden category have exceptionally high medication costs?

Within each burden category, are there patients whose costs are exceptionally high compared with their peers?

```python
def count_cost_outliers(group):
    q1 = group['total_medication_cost'].quantile(0.25)
    q3 = group['total_medication_cost'].quantile(0.75)
    iqr = q3 - q1
    upper_limit = q3 + 1.5 * iqr
    
    return pd.Series({
        'patients': len(group),
        'upper_cost_limit': round(upper_limit, 2),
        'outlier_patients': (group['total_medication_cost'] > upper_limit).sum()
    })

category_cost_outliers = (
    patient_medication_burden
    .groupby('medication_burden_category')
    .apply(count_cost_outliers)
)

category_cost_outliers
```

```python
"""Conceptually, Pandas does:

Low patients
   ↓
calculate Q1 → Q3 → IQR → upper limit → count outliers

Moderate patients
   ↓
calculate Q1 → Q3 → IQR → upper limit → count outliers

Polypharmacy patients
   ↓
calculate Q1 → Q3 → IQR → upper limit → count outliers"""
```

Here, each category has its own benchmark.

Low: 320 patients exceed ₹266,804.98.
Moderate: 100 patients exceed ₹613,508.71.
Polypharmacy: 17 patients exceed ₹1,203,774.43.

```python
category_cost_outliers['outlier_pct'] = (
    category_cost_outliers['outlier_patients']
    / category_cost_outliers['patients']
    * 100
).round(2)

category_cost_outliers
```

Key finding

Polypharmacy has the highest proportion of unusually high-cost patients: 5.18%.

Approximately 4–5% of patients in each medication-burden category were identified as high-cost outliers using category-specific IQR thresholds, with the highest proportion observed in the polypharmacy group (5.18%).

---examine which burden category contributes the largest share of the patient population and total spending. This will combine patient distribution + spending distribution, rather than another correlation.

```python
burden_category_summary = (
    patient_medication_burden
    .groupby('medication_burden_category')
    .agg(
        patients=('total_medication_cost', 'size'),
        total_cost=('total_medication_cost', 'sum'),
        avg_cost=('total_medication_cost', 'mean')
    )
    .round(2)
)

burden_category_summary
```
